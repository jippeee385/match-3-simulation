```python
import numpy as np
import pandas as pd

from match3.game.level import Level, Objective, ObjectiveType
from match3.players.random_player import RandomPlayer
from match3.simulation.simulator import simulate_games
```

```python
easy_level = Level(
    name="Easy",
    moves=25,
    objective=Objective(
        type=ObjectiveType.COLLECT,
        target=10,
        candy_type=0,
    ),
)

medium_level = Level(
    name="Medium",
    moves=15,
    objective=Objective(
        type=ObjectiveType.COLLECT,
        target=20,
        candy_type=0,
    ),
)

hard_level = Level(
    name="Hard",
    moves=10,
    objective=Objective(
        type=ObjectiveType.COLLECT,
        target=25,
        candy_type=0,
    ),
)
```

```python
for level in [easy_level, medium_level, hard_level]:
    rng = np.random.default_rng(42)
    player = RandomPlayer(rng)

    results = simulate_games(
        level=level,
        player=player,
        n_games=1000,
        rng=rng,
    )

    results_df = pd.DataFrame(results)

    print(
        level.name,
        results_df["won"].mean(),
    )
```
