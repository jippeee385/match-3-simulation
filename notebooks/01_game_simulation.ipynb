```python
import numpy as np
import pandas as pd

from match3.game.level import Level, Objective, ObjectiveType
from match3.players.random_player import RandomPlayer
from match3.simulation.simulator import simulate_game
```

```python
level = Level(
    name="Collect 10 Red",
    moves=25,
    objective=Objective(
        type=ObjectiveType.COLLECT,
        target=10,
        candy_type=0,
    ),
)

print(f"Level: {level.name}")
print(f"Moves: {level.moves}")
print(f"Objective: Collect {level.objective.target} of candy type "
      f"{level.objective.candy_type}")
```

```python
rng = np.random.default_rng(42)

player = RandomPlayer(rng)

result = simulate_game(
    level,
    player,
    rng=rng,
)

print("\nSingle game result:")
print(result)
```

```python
result_df = pd.DataFrame([result])

print("\nResult as DataFrame:")
display(result_df)
```
